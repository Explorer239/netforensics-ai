## 7. Missing Value Analysis

Missing values can reduce the quality and reliability of the network troubleshooting dataset.

This step checks each column for missing values.

```python
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

## 8. Total Missing Values

This step calculates the total number of missing values present in the complete dataset.

```python
total_missing = df.isnull().sum().sum()

print("Total missing values:", total_missing)

if total_missing == 0:
    print("No missing values found.")
else:
    print("Some missing values are present.")
```

## 9. Duplicate Record Analysis

Duplicate records can reduce the quality of the dataset and may affect the reliability of the NetForensics AI system.

This step checks for completely duplicated rows.

```python
duplicate_rows = df.duplicated().sum()

print("Duplicate rows:", duplicate_rows)

if duplicate_rows == 0:
    print("No duplicate records found.")
else:
    print("Duplicate records found.")
```

## 10. Case ID Validation

Each network troubleshooting case should have a unique Case ID.

This step checks whether any Case IDs are repeated.

```python
duplicate_case_ids = df["Case_ID"].duplicated().sum()

print("Duplicate Case IDs:", duplicate_case_ids)

if duplicate_case_ids == 0:
    print("All Case IDs are unique.")
else:
    print("Duplicate Case IDs found.")
```

## 11. Category/Protocol Distribution

The NetForensics AI dataset contains cases related to different networking protocols and technologies.

This analysis shows the number of cases associated with each protocol.

```python
protocol_count = df["Category"].value_counts()

print("Protocol Distribution:")
print(protocol_count)
```

## 12. OSI Layer Distribution

Network faults can occur at different layers of the OSI model.

This analysis shows how the troubleshooting cases are distributed across different OSI layers.

```python
osi_count = df["OSI_Layer"].value_counts()

print("OSI Layer Distribution:")
print(osi_count)
```

## 13. Severity Distribution

Each network fault case is categorized based on severity.

The severity level helps prioritize network troubleshooting and corrective actions.

```python
severity_count = df["Severity"].value_counts()

print("Severity Distribution:")
print(severity_count)
```

## 14. Empty String Analysis

Some cells may not contain missing values in the form of NaN but may contain blank or empty strings.

This step checks for empty string values in each column.

```python
empty_count = 0

for column in df.columns:
    count = df[column].astype(str).str.strip().eq("").sum()
    empty_count = empty_count + count

    print(column, ":", count)

print("\nTotal empty values:", empty_count)
```

## 15. Dataset Validation Summary

This section provides a summary of all important validation checks performed on the NetForensics AI dataset.

```python
print("--------------------------------------------")
print("NETFORENSICS AI DATASET VALIDATION SUMMARY")
print("--------------------------------------------")

print("Total Cases:", rows)
print("Total Columns:", columns)
print("Missing Values:", total_missing)
print("Duplicate Rows:", duplicate_rows)
print("Duplicate Case IDs:", duplicate_case_ids)
print("Empty Values:", empty_count)

print("                                              ")
if total_missing == 0 and duplicate_rows == 0 and duplicate_case_ids == 0:
    print("Dataset is valid and ready for the next phase.")
else:
    print("Dataset needs to be reviewed.")
```
