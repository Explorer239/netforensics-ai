## 4. Rule-Based Fault Detection Approach

A rule-based system uses predefined conditions to identify possible network faults.

The system checks the symptom and evidence of each network troubleshooting case for specific keywords or patterns.

When a condition matches, the system provides:

- Detected fault
- Recommended troubleshooting command
- Suggested fix

This approach is useful for identifying common and well-defined network problems.

## 5. Create the Rule-Based Detection Function

The following function analyzes the available symptom and evidence.

Simple rules are created for common network faults such as:

- Interface shutdown
- IP addressing problems
- Default gateway problems
- VLAN issues
- Trunk issues
- Routing problems
- DNS problems
- DHCP problems
- SSH problems
- STP problems

If no rule matches the available information, the system returns "Unknown Fault".

```python
def detect_fault(symptom, evidence):

    # Convert input into lowercase
    text = str(symptom).lower() + " " + str(evidence).lower()

    # Interface shutdown
    if "administratively down" in text:
        return "Interface Shutdown", "show ip interface brief", "Use the no shutdown command"

    # Duplicate IP address
    elif "duplicate ip" in text or "ip address conflict" in text:
        return "Duplicate IP Address", "ipconfig", "Configure a unique IP address"

    # Wrong default gateway
    elif "default gateway" in text and ("incorrect" in text or "wrong" in text):
        return "Default Gateway Mismatch", "ipconfig", "Configure the correct default gateway"

    # VLAN problem
    elif "vlan" in text and ("missing" in text or "wrong" in text):
        return "VLAN Configuration Issue", "show vlan brief", "Configure the correct VLAN"

    # Trunk problem
    elif "trunk" in text and ("down" in text or "mismatch" in text):
        return "Trunk Configuration Issue", "show interfaces trunk", "Configure the trunk correctly"

    # Routing problem
    elif "route" in text and ("missing" in text or "not found" in text):
        return "Routing Issue", "show ip route", "Configure the correct route"

    # DNS problem
    elif "dns" in text or "nslookup" in text:
        return "DNS Configuration Issue", "nslookup", "Check the DNS server configuration"

    # DHCP problem
    elif "dhcp" in text or "169.254" in text:
        return "DHCP Configuration Issue", "ipconfig /all", "Check the DHCP configuration"

    # SSH problem
    elif "ssh" in text and ("failed" in text or "refused" in text):
        return "SSH Configuration Issue", "show running-config", "Check SSH configuration"

    # STP problem
    elif "blocking" in text and "port" in text:
        return "STP Port Issue", "show spanning-tree", "Check STP configuration"

    # Ping failure
    elif "100% loss" in text or "request timed out" in text:
        return "Connectivity Issue", "ping", "Check IP addressing and network connectivity"

    else:
        return "Unknown Fault", "Check network configuration", "Manual investigation required"
```

## 6. Test the Rule-Based Detection Function

Before applying the rules to the complete dataset, the function is tested using a sample network fault.

This helps verify that the rule-based detection function is working correctly.

```python
fault, command, fix = detect_fault(
    "PC cannot communicate with another network",
    "Interface is administratively down"
)

print("Detected Fault:", fault)
print("Recommended Command:", command)
print("Suggested Fix:", fix)
```

## 7. Apply the Rule-Based System to the Dataset

The rule-based detection function is now applied to each network troubleshooting case.

The system analyzes the symptom and evidence of every case and stores:

- Detected fault
- Recommended command
- Suggested fix

```python
detected_faults = []
recommended_commands = []
suggested_fixes = []

for index, row in df.iterrows():

    symptom = row["Symptom"]
    evidence = row["Evidence"]

    fault, command, fix = detect_fault(symptom, evidence)

    detected_faults.append(fault)
    recommended_commands.append(command)
    suggested_fixes.append(fix)
```

## 8. Store Rule-Based Results

The detected results are added to the dataset as new columns.

This makes it possible to compare the rule-based diagnosis with the expected network fault.

```python
df["Rule Detected Fault"] = detected_faults
df["Rule Recommended Command"] = recommended_commands
df["Rule Suggested Fix"] = suggested_fixes

print("Rule-based results added successfully.")
```

## 9. Display Rule-Based Detection Results

This step displays the original troubleshooting information along with the results generated by the rule-based system.

```python
df[[
    "Case_ID",
    "Expected_Fault",
    "Rule Detected Fault",
    "Rule Recommended Command",
    "Rule Suggested Fix"
]].head(10)
```

## 10. Compare Rule-Based Detection with Expected Fault

The detected fault is compared with the expected fault available in the original dataset.

This comparison helps evaluate whether the rule-based system correctly identifies the network problem.

```python
df["Rule Match"] = (
    df["Expected_Fault"].str.lower() ==
    df["Rule Detected Fault"].str.lower()
)

print("Rule Matching Completed.")
```

## 11. Calculate Rule-Based Detection Accuracy

The accuracy is calculated by finding the percentage of cases where the rule-based detected fault matches the expected fault.

```python
correct_cases = df["Rule Match"].sum()
total_cases = len(df)

accuracy = (correct_cases / total_cases) * 100

print("Correctly Detected Cases:", correct_cases)
print("Total Cases:", total_cases)
print("Rule-Based Detection Accuracy:", round(accuracy, 2), "%")
```

## 12. Display Cases with Correct Detection

This step displays cases where the rule-based system successfully detected the expected network fault.

```python
correct_results = df[df["Rule Match"] == True]

correct_results[[
    "Case_ID",
    "Symptom",
    "Expected_Fault",
    "Rule Detected Fault"
]]
```

## 13. Display Cases Requiring Further Analysis

Some network faults cannot be identified using simple predefined rules.

These cases are displayed for further analysis and will later be handled using AI-assisted troubleshooting.

```python
incorrect_results = df[df["Rule Match"] == False]

incorrect_results[[
    "Case_ID",
    "Symptom",
    "Expected_Fault",
    "Rule Detected Fault"
]]
```
